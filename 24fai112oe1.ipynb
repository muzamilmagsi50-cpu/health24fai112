```python
import pandas as pd
df = pd.read_csv("diabetes.csv")
```

```python
df.head(10)
```

```python
df.info()
```

```python
# Checking for the missing values in the dataset
df.isnull().sum()
```

```python
df.drop("Outcome", axis=1)
```

```python
df["Outcome"]
```

```python
#now separating for outputs
X = df.drop("Outcome", axis=1)
y = df["Outcome"]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=56
)

# this is used for scaling of data
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier (random_state=56)
```

```python
model.fit (X_train, y_train)
```

```python
#using predictions
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

# for accuracy score of the test/training

```

```python
print (accuracy_score(y_test, y_pred))
```

```python
print (accuracy_score(y_test, y_pred) *100,"%")
```

```python
from sklearn.metrics import confusion_matrix
# imporing for simple matrix output first
```

```python
cm= confusion_matrix(y_test, y_pred)
```

```python
print(cm)
```

```python
# for training accuracy
train_pred = model.predict(X_train)
train_accuracy = accuracy_score(y_train, train_pred)
print("SVM Training Accuracy:", round(train_accuracy * 100, 2),"%")
```

```python
# using tree classifier to check overfitting
tree_model = DecisionTreeClassifier(random_state=56)
tree_model.fit(X_train, y_train)
```

```python
tree_pred = tree_model.predict(X_train)
tree_train_accuracy = accuracy_score(y_train, tree_pred)
print("Decision Tree Training Accuracy:",round(tree_train_accuracy * 100, 2),"%")
```

```python
# reducing overfitting
tree_model = DecisionTreeClassifier(max_depth=4, random_state=56)
tree_model.fit(X_train, y_train)
```

```python
tree_pred = tree_model.predict(X_test)
print("Decision Tree Accuracy:", round(accuracy_score(y_test, tree_pred) * 100, 2),"%")
```

```python
tree_train = tree_model.predict(X_train)
print("Decision Tree Training Accuracy:", round(accuracy_score(y_train, tree_train) * 100, 2),"%")
```
